```python
from faker import Faker 
import random 
fake = Faker("en_IN")
random.seed(42)
Faker.seed(42)
print("Faker initialized successfully!")
```

```python
NUM_CUSTOMERS = 1000
customer_segments = [
    "Premium",
    "GOld",
    "Silver",
    "Regular"
]
occupations = [
    "Software Engineer",
    "Doctor",
    "Teacher",
    "Business",
    "Student",
    "Lawyer",
    "Designer",
    "Manager",
    "Data Analyst",
    "Accountant"
]
marital_status = [
    "Single",
    "Married"
]
print("Configuration Loaded Successfully!")
print(f"Customer to Generate : {NUM_CUSTOMERS}")
```

```python
def generate_customer(customer_id):

    customer = {
        "customer_id": customer_id,
        "first_name": fake.first_name(),
        "last_name": fake.last_name(),
        "gender": random.choice(["Male", "Female"]),
        "age": random.randint(18, 70),
        "email": fake.email(),
        "phone": fake.phone_number(),
        "city": fake.city(),
        "state": fake.state(),
        "country": "India",
        "signup_date": fake.date_between(start_date="-5y", end_date="today"),
        "customer_segment": random.choice(customer_segments),
        "annual_income": random.randint(200000, 3000000),
        "marital_status": random.choice(marital_status),
        "occupation": random.choice(occupations),
        "loyalty_points": random.randint(0, 10000),
        "is_active": random.choice([True, False])
    }

    return customer
```

```python
customer = generate_customer(1)

customer
```

```python
customers = []

for customer_id in range(1, NUM_CUSTOMERS + 1):
    customers.append(generate_customer(customer_id))

print(f"Total Customers Generated: {len(customers)}")
```

```python
customers_df = pd.DataFrame(customers)
customers_df.head()
```

```python
segment_counts = customers_df["customer_segment"].value_counts()

plt.figure(figsize=(8,5))
segment_counts.plot(kind="bar")

plt.title("Customer Segment Distribution")
plt.xlabel("Customer Segment")
plt.ylabel("Number of Customers")

plt.show()
```

```python
gender_counts = customers_df["gender"].value_counts()

plt.figure(figsize=(6,6))
gender_counts.plot(kind="pie", autopct="%1.1f%%")

plt.title("Gender Distribution")
plt.ylabel("")

plt.show()
```
