```python
sales = (
    orders.merge(customers, on="customer_id", how="inner")
          .merge(products, on="product_id", how="inner")
)

sales.head()
```

```python
sales.shape
```

```python
total_revenue = sales["total_amount"].sum()
print(f"Total Revenue: ₹{total_revenue:,.2f}")
```

```python
print("Total Customers:", sales["customer_id"].nunique())
```

```python
print("Total Quantity Sold:", sales["quantity"].sum())
```

```python
print(f"Average Order Value: ₹{sales['total_amount'].mean():,.2f}")
```

```python
top_customers = (
    sales.groupby(["customer_id", "first_name", "last_name"], as_index=False)
         .agg(
             Revenue=("total_amount", "sum"),
             Orders=("order_id", "count")
         )
         .sort_values("Revenue", ascending=False)
         .head(10)
)

top_customers
```

```python
top_products = (
    sales.groupby(["product_id", "product_name"], as_index=False)
         .agg(
             Quantity_Sold=("quantity", "sum"),
             Revenue=("total_amount", "sum")
         )
         .sort_values("Revenue", ascending=False)
         .head(10)
)

top_products
```
